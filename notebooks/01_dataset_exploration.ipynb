```python
import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from pathlib import Path

pd.set_option("display.max_columns", None)
pd.set_option("display.max_rows", 100)

print("Environment ready!")
```

```python
PROJECT_ROOT = Path.cwd().parent

DATA_DIR = PROJECT_ROOT / "data" / "raw"

DATA_FILE = DATA_DIR / "dataSet_withFeatures.csv"
LABEL_FILE = DATA_DIR / "dataSet_withFeatures_labels.csv"

print("Data file:", DATA_FILE)
print("Label file:", LABEL_FILE)

print("\nData file exists:", DATA_FILE.exists())
print("Label file exists:", LABEL_FILE.exists())
```

```python
X_raw = pd.read_csv(
    DATA_FILE,
    header=None
)

y_raw = pd.read_csv(
    LABEL_FILE,
    header=None
)

print("X shape:", X_raw.shape)
print("y shape:", y_raw.shape)
```

```python
sample = X_numeric.iloc[0, :414].to_numpy()

tactile_sample = sample.reshape(6, 69)

print("Original tactile vector shape:", sample.shape)
print("Reshaped tactile sample:", tactile_sample.shape)
```

```python
fig, axes = plt.subplots(6, 1, figsize=(12, 12), sharex=True)

for i in range(6):
    axes[i].plot(tactile_sample[i])
    axes[i].set_ylabel(f"Taxel {i+1}")
    axes[i].grid(True, alpha=0.3)

axes[-1].set_xlabel("Time step")

plt.suptitle("Raw Tactile Sensor Readings for One Robot Step")
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(12, 6))

for i in range(6):
    plt.plot(
        tactile_sample[i],
        label=f"Taxel {i+1}"
    )

plt.title("Six Tactile Channels for One Robot Step")
plt.xlabel("Time step")
plt.ylabel("Sensor value")
plt.legend()
plt.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
```

```python
X = X_raw.iloc[:, :447].copy()

tactile_data = X.iloc[:, :414].copy()
feature_data = X.iloc[:, 414:447].copy()

print("Complete input:", X.shape)
print("Tactile data:", tactile_data.shape)
print("Engineered features:", feature_data.shape)
print("Labels:", y_raw.shape)
```

```python
sample_idx = 0

sample_tactile = tactile_data.iloc[sample_idx].to_numpy()

sample_tactile = sample_tactile.reshape(6, 69)

print("Shape:", sample_tactile.shape)
```

```python
fig, axes = plt.subplots(
    6, 1,
    figsize=(12, 12),
    sharex=True
)

for i in range(6):
    axes[i].plot(sample_tactile[i])
    axes[i].set_ylabel(f"Taxel {i+1}")
    axes[i].grid(alpha=0.3)

axes[-1].set_xlabel("Time step")

plt.suptitle(
    f"Tactile Signal for Sample {sample_idx} "
    f"({labels.iloc[sample_idx]})"
)

plt.tight_layout()
plt.show()
```

```python
fig, axes = plt.subplots(
    4, 1,
    figsize=(14, 12),
    sharex=True
)

unique_labels = labels.unique()

for ax, terrain in zip(axes, unique_labels[:4]):

    idx = labels[labels == terrain].index[0]

    signal = tactile_data.iloc[idx].to_numpy().reshape(6, 69)

    for channel in range(6):
        ax.plot(
            signal[channel],
            label=f"T{channel+1}"
        )

    ax.set_title(f"Terrain: {terrain}")
    ax.grid(alpha=0.3)
    ax.legend(ncol=6)

axes[-1].set_xlabel("Time step")

plt.tight_layout()
plt.show()
```

```python
fig, axes = plt.subplots(
    4, 1,
    figsize=(14, 12),
    sharex=True
)

for ax, terrain in zip(axes, unique_labels[4:]):

    idx = labels[labels == terrain].index[0]

    signal = tactile_data.iloc[idx].to_numpy().reshape(6, 69)

    for channel in range(6):
        ax.plot(
            signal[channel],
            label=f"T{channel+1}"
        )

    ax.set_title(f"Terrain: {terrain}")
    ax.grid(alpha=0.3)
    ax.legend(ncol=6)

axes[-1].set_xlabel("Time step")

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(12, 5))

for terrain in labels.unique():

    idx = labels[labels == terrain].index[0]

    signal = tactile_data.iloc[idx].to_numpy().reshape(6, 69)

    plt.plot(
        signal[5],
        label=terrain
    )

plt.title("Shear Taxel Signal Across Terrains")
plt.xlabel("Time step")
plt.ylabel("Shear sensor value")
plt.legend()
plt.grid(alpha=0.3)

plt.tight_layout()
plt.show()
```
